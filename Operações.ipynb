```guile
(define function-data
  '((0 0 1 1)
    (0 0 1 2)
    (1 1 0 0)
    (1 2 3 0)))

(define (sum func1 func2)
  (let ([a (+ (car func1) (car func2))]
        [b (+ (cadr func1) (cadr func2))]
        [c (+ (cadr (cdr func1)) (cadr (cdr func2)))]
        [d (+ (cadr (cdr (cdr func1))) (cadr (cdr (cdr func2))))])
  `(,a ,b ,c ,d)))

(define (sub func1 func2)
  (let ([a (- (car func1) (car func2))]
        [b (- (cadr func1) (cadr func2))]
        [c (- (cadr (cdr func1)) (cadr (cdr func2)))]
        [d (- (cadr (cdr (cdr func1))) (cadr (cdr (cdr func2))))])
  `(,a ,b ,c ,d)))

(define (mul func1 func2)
  (let ([a (* (car func1) (car func2))]
        [b (* (cadr func1) (cadr func2))]
        [c (* (cadr (cdr func1)) (cadr (cdr func2)))]
        [d (* (cadr (cdr (cdr func1))) (cadr (cdr (cdr func2))))])
  `(,a ,b ,c ,d)))

(define (div func1 func2)
  (let ([a (if (= (car func2) 0)
               '(error)
               (/ (car func1) (car func2)))]
        [b (if (= (cadr func2) 0)
               '(error)
               (/ (cadr func1) (cadr func2)))]
        [c (if (= (cadr (cdr func2)) 0)
               '(error)
               (/ (cadr (cdr func1)) (cadr (cdr func2))))]
        [d (if (= (cadr (cdr (cdr func2))) 0)
               '(error)
               (/ (cadr (cdr (cdr func1))) (cadr (cdr (cdr func2)))))])
  `(,a ,b ,c ,d)))
```
