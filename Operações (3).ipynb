```guile
(define (coef expr)
  (if (and (pair? expr)
           (eq? (car expr) '*)
           (number? (cadr expr)))
      (cadr expr)
      1))

(define (var expr)
  (cond ((and (pair? expr)
              (eq? (car expr) '*)
              (number? (cadr expr)))
         (var (caddr expr)))
        ((and (pair? expr)
              (eq? (car expr) '**))
         (cadr expr))
        (else expr)))

(define (base expr)
  (if (and (pair? expr) (eq? (car expr) '**))
      (cadr expr)
      expr))

(define (expoente expr)
  (cond ((and (pair? expr)
              (eq? (car expr) '*)
              (number? (cadr expr)))
         (expoente (caddr expr)))
        ((and (pair? expr)
              (eq? (car expr) '**))
         (caddr expr))
        (else 1)))

(define (sum x y) 
  (cond ((and (number? x) (number? y)) (+ x y))
        ((eq? (var x) (var y))
         (list '* (+ (coef x)
                 (coef y))
                 (var x)))
        (else
         (list '+ x y))))

(define (sub x y) 
  (cond ((and (number? x) (number? y)) (- x y))
        ((eq? (var x) (var y))
         (list '* (- (coef x)
                 (coef y))
                 (var x)))
        (else
         (list '- x y))))

(define (mul x y) 
  (cond ((and (number? x) (number? y)) (* x y))
        ((eq? (var x) (var y))
         (list '* (* (coef x)
                 (coef y))
                 (list '** (var x)
                       (+ (expoente x)
                          (expoente y)))))
        (else
         (list '* x y))))

(define (div x y) 
  (cond 
        ((equal? x y) 1)
        ((and (number? x) (number? y)) (/ x y))
        ((eq? (var x) (var y))
         (list '/ (/ (coef x)
                 (coef y))
                 (list '** (var x)
                       (- (expoente x)
                          (expoente y)))))
        (else
         (list '/ x y))))
```
